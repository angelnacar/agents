## Investigación MCP Jira

```python
import os
from dotenv import load_dotenv
from agents import Agent, Runner, trace, Tool
from agents.mcp import MCPServerStdio
from IPython.display import Markdown, display
from datetime import datetime
import asyncio
load_dotenv(override=True)

jira_env = {"JIRA_API_TOKEN": os.getenv("JIRA_API_TOKEN")}
jira_username = {"JIRA_USERNAME": os.getenv("JIRA_USERNAME")}


mcp_params = {
    "command": "uvx",
    "args": ["mcp-atlassian"],
    "env": {
        "JIRA_URL": "https://giss.atlassian.net",
        "JIRA_USERNAME": jira_username["JIRA_USERNAME"],
        "JIRA_API_TOKEN": jira_env["JIRA_API_TOKEN"]
      }
}

async with MCPServerStdio(params=mcp_params, client_session_timeout_seconds=30) as atlassian_mcp:
  agent = Agent(
        name="Jira Agent",
        instructions="""
            Eres un asistente experto en gestión de proyectos Jira.
            Ayudas a gestionar issues, sprints y proyectos de forma eficiente.
        """,
        mcp_servers=[atlassian_mcp],
        model="gpt-4o-mini",
    )


  result = await Runner.run(agent, "Lista los issues abiertos del proyecto IVFIEQAA", max_turns=30)
  display(Markdown(f"**Respuesta del agente:** {result}"))
```
